# IFC Sample Model Analysis — Portfolio

This portfolio presents a structured exploration of Industry Foundation Classes (IFC) data, combining semantic, geometric, network-graph, and spatial analysis within a reproducible computational workflow.

The objective is to examine how building information models can be systematically queried, interpreted, and transformed into analysis-ready datasets for downstream applications such as GIS integration, validation, and exposure modeling.

It was split out of a single monolithic notebook into one standalone notebook per analytical step — each notebook re-downloads the sample model and runs independently, so you can open any one of them in Colab without running the others first.

## Model

The IFC sample model `AC20-FZK-Haus.ifc` is provided as part of the KIT IFC example dataset, developed by the Institute for Automation and Applied Informatics (IAI) at Karlsruhe Institute of Technology (KIT). It is distributed for unrestricted use and commonly referenced in IFC tutorials, including IfcOpenShell documentation and sample workflows.

## Pipeline

```
01 Inventory & Properties   — what elements exist, what they know about themselves
        │
02 Geometry                 — shape extraction, mesh preview
        │
03 Spatial Hierarchy & Graphs — the model as a network: containment,
        │                       openings, materials, classification;
        │                       interactive + static views, graph analytics
04 Classification           — element counts by class / predefined type
        │
05 GIS Extraction           — footprints, georeferencing, GeoJSON export
        │
06 Analysis & Validation    — data-quality checks, validation report
```

## Local setup

Each notebook installs its own dependencies via `!pip install` on first run, matching how they'd be opened fresh in Colab. To run them locally instead (e.g. in Jupyter or VS Code), install everything up front:

```bash
pip install -r requirements.txt
```

The two `google.colab.*` calls (the interactive inventory sheet in 01, and the file-download cells in 03/05) only run in Colab — running them locally raises `ModuleNotFoundError: No module named 'google'`. That's expected; every other cell is Colab/local portable.

## Notebooks

- [01 · Inventory & Properties](./01_inventory_properties.ipynb) — element inventory table and property-set inspection.
- [02 · Geometry](./02_geometry.ipynb) — shape extraction for a wall and a slab, a triangulated mesh preview.
- [03 · Spatial Hierarchy & Graphs](./03_spatial_graphs.ipynb) — the model as a network graph: spatial containment, openings, materials, and classification, with an interactive PyVis view, a static hierarchy render, and graph analytics (degree distribution, centrality, connected components).
- [04 · Classification](./04_classification.ipynb) — element counts by class, predefined type, and load-bearing status.
- [05 · GIS Extraction](./05_gis_extraction.ipynb) — 2D footprints, approximate georeferencing, GeoJSON export.
- [06 · Analysis & Validation](./06_validation.ipynb) — data completeness, spatial-structure, geometry-quality, and semantic consistency checks, consolidated into a validation report.

### What changed from the original single notebook
- **Fixed a bug**: the original spatial-hierarchy graph was built correctly, then immediately overwritten by a second, thinner graph before half its relationships were re-added — the two are now merged into one correct build.
- **Deduplicated** a mesh-extraction helper that was defined twice.
- **Added** a Project→Site→Building→Storey→Elements hierarchy graph with a static layered render, a material-association graph, a classification-association graph, and graph analytics.
- **Expanded** the validation section from one implemented check to the full checklist (completeness, spatial structure, geometry quality, semantic consistency, georeferencing, exposure), consolidated into a single validation report.
